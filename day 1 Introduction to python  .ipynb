```python
# 14. WAP to create final bill for medium biryanis ordered each of which costs 200 Rs.
#     where delievery platform is giving 20% discount and 5% tax applicable on final Bill. 


# Taking Input from User 
q = int(input("  Enter Quantity : ") ) 

# Expression 
amt =  200 * q                        # Actual amount 
dis =  20 / 100 * amt                 # discount in Rupees
bill = amt - dis                      # Discounted amount 
tax =  5 / 100 * bill                 # Tax applicable 

# Display finall Biil 

print("\n\n")
print("---------------------------- ~ Final Bill ~ ---------------------------- ")
print()
print(" \t\t  Quantity Ordered       :    ", q)
print(" \t\t  Total Bill             :    ", amt)
print(" \t\t  Discount               :    ", dis)
print(" \t\t  Amount after Discount  :    ", bill)
print(" \t\t  Tax Applicable         :    ", tax)
print("-----------------------------------------------------------------")
print(" \t\t  Final Bill             :    ", bill + tax)
```

```python
# 17. WAP to to calculate CAGR. 

# Taking Input from User 

b = int(input(" Enter starting amount : "))
e = int(input(" Enter ending   amount : "))
t = float(input(" Enter Tenure Years  : "))

# Expression 

CAGR = (( e / b ) ** (1 / t)) - 1 

# Output 
print("Compound Annual Growth Rate (CAGR) : ", round(CAGR * 100, 2) , "%")
```

```python
# 18.  WAP to calculate distance from two points. 

# Taking Input from User 
x1 = float(input(" Enter x1 : "))
x2 = float(input(" Enter y1 : "))
y1 = float(input(" Enter x2 : "))
y2 = float(input(" Enter y2 : "))

# Expression 
d = (  (y2 - y1) ** 2  + (x2 - x1) ** 2   ) ** (1/2)

# Display Output
print(" Distance : ", d )
```

```python
# 19. WAP to display Salary slip of an employee for following heads .. 
#     tot = bs + da + ta + hra + fixed 
#     da = 60%       ta = 30%        hra = 40%         fixed = 5000


# Taking Input from User 
bs = int(input(" Enter Basic Salary  : "))

# Expressions 
da  = 60 / 100 * bs
ta  = 30 / 100 * bs
hra = 40 / 100 * bs 
f   = 5000

tot = bs + da + ta + hra + f

# Display Output 
print(" ------------------ ~ Salary Slip ~  --------------------- ")
print()
print(" \t \t Basic Salary            : ", bs )
print(" \t \t Dearness Allowance      : ", da )
print(" \t \t Travelling Allowance    : ", ta )
print(" \t \t House Rent Allowance    : ", hra )
print(" \t \t Fixed Component         : ", f )
print("------------------------------------------------------------")
print(" \t \t Total Salary            : ", tot )
```
