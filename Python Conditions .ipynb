```python
# 2. WAP to calculate BMI and interpret it. 
"""
Metric Formul = weight / ht square 

Underweight: BMI less than 18.5
Healthy Weight: BMI 18.5 to 24.9
Overweight: BMI 25.0 to 29.9
Obesity: BMI 30.0
"""

# Taking Input from User 
wt = int(input(" Enter Your Weight in Kg : "))
ht = float(input(" Enter Your Height in meters : "))

# Expression 
BMI = wt / ht ** 2 

# Interpretation 
if BMI <= 18.5 : 
    print(" Underweight ")
elif BMI <= 24.9  :
    print(" Healthy Weight ")
elif BMI <= 29.9 : 
    print("Overweight  ")
else : 
    print(" Obesity ")
```

```python
# 3. During a festive season following offers are launched by a retail firm 
#    0 - 1000  : 10%
#    upto 2000 : 20%
#    upto 3000 : 30%
#    upto 4000 : 40%
#    >4000     : 50%
#    WAP to display final bill. 

# Taking Input from User 
b = int(input(" Enter Bill Amount : "))

# Conditions 
if b <= 1000 : 
    d = 10
elif b <= 2000 : 
    d = 20 
elif b <= 3000 : 
    d = 30
elif b <= 4000 : 
    d = 40
else : 
    d = 50
    
# Expression 
amt_d = d / 100 * b

# Display Bill 
print("\n\n")
print(" Total Bill      : ", b)
print(" % Discount      : ", d, "%")
print(" Discount Amount : ", amt_d)    
print(" Final Bill      : ", b - amt_d)
```

```python
# 4. WAP to implement FD calculator for SBI. 

"""                                      Public      Senior Citizen (60+)
7 days to 45 days                       3.05         3.55
46 days to 179 days                     4.90         5.40
180 days to 210 days                    5.65         6.15
211 days to less than 1 year            5.90         6.40
1 Year to less than 2 years             6.25         6.75
2 years to less than 3 years            6.40         6.90
3 years to less than 5 years            6.30         6.80
5 years and up to 10 years              6.05         7.05*
"""
```

```python
# Taking Input from User 

P  =  int(input(" Enter Amount : "))
N  =  int(input(" Enter Tenure in Days : "))

# Conditions 
if N >= 7 and N <= 45 : 
    R = 3.05
elif N <= 179 : 
    R = 4.9
elif N <= 210 : 
    R = 5.65
elif N <= (1 * 365) : 
    R = 5.90
elif N <= (2 * 365) :
    R = 6.25
elif N <=  (3 * 365) : 
    R = 6.40
elif N <= (5 * 365) : 
    R = 6.30
elif N <= (10 * 365) : 
    R = 6.05
    
# Expression 
A = P * (1 + R / 100 ) ** (N / 365)

# Display Output 
print("\n\n")
print(" Invested Amount  : ", P)
print(" Tenure           : ", N)
print(" Rate of Interest : ", R , "%")
print(" Final Amount     : ", A)
print(" Interest Earned  : ", A - P)
```

```python
"""
Up to ₹4 Lakh:          Nil
₹4 Lakh to ₹8 Lakh:     5%
₹8 Lakh to ₹12 Lakh:    10%
₹12 Lakh to ₹16 Lakh:   15%
₹16 Lakh to ₹20 Lakh:   20%
₹20 Lakh to ₹24 Lakh:   25%
Above ₹24 Lakh:         30%
"""

# Taking input from User 
inc = int(input(" Enter Income from all Sources : "))

# Conditions 
if inc <= 400000 : 
    tax = 0
elif inc <= 800000 : 
    tax = 5 / 100 * (inc - 400000 )
elif inc <= 1200000 : 
    tax = 20000 + 10 / 100 * (inc - 800000 )
elif inc <= 1600000 : 
    tax = 60000 + 15 / 100 * (inc - 1200000)
elif inc <= 2000000 : 
    tax = 120000 + 20 / 100 * (inc - 1600000)
elif inc <= 2400000 : 
    tax = 200000 + 25 / 100 * (inc - 2000000)
else : 
    tax = 300000 + 30 / 100 * (inc - 2400000)
    

# Output
print(" Tax Liability : ", tax)
```

```python
# WAP to display Salary Slip. 

#       BS         DA     TA       HRA       Fixed
#      <10000      70%    50%      50%       8000
#      <20000      75%    60%      60%       15000
#      <30000      80%    70%      70%       20000
#      >30000      80%    30%      80%       25000
```

```python
#  Taking Input from User 

bs = int(input(" Enter Basic Salary : "))

if bs <= 10000 : 
    da, ta, hra, f =70, 50, 50, 8000
elif bs <= 20000 : 
    da, ta, hra, f =75, 60, 60, 15000
elif bs <= 30000 : 
    da, ta, hra, f =80, 70, 70, 20000
else : 
    da, ta, hra, f =80, 30, 80, 25000
    
# Calculations 
DA = da / 100 * bs
TA = ta / 100 * bs
HRA = hra / 100 * bs
tot = bs + DA + TA+ HRA + f
# Display Salary slip 

print("\n\n")
print("\t\t Basic Salary              : ", bs)
print("\t\t Dearness Allowance        : ", DA)
print("\t\t Travelling Allowance      : ", TA)
print("\t\t House Rent Allowance      : ", HRA)
print("\t\t Fixed Pay                 : ", f)
print("\t\t Total Salary              : ",tot )
```
