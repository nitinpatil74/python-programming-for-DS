```python
# copy()

s1 = { 11,22,33,44,55}
s2 = s1.copy()

print(s1)
print(s2)
```

```python
# clear()

s1 = { 11,22,33,44,55}
s1.clear()
print(s1)
```

```python
# 'isdisjoint'

s1 = {11,22,33}
s2 = {77,88,99}

s1.isdisjoint(s2)
```

```python
# 'issubset' ,  'issuperset' 

s1 = { 11,22,33,44,55,66,77,88 }     # Superset
s2 = { 44,66,88}                     # Subset

print(s1.issuperset(s2))
print(s2.issubset(s1))
```

```python
# 'union'

s1 = {11,22,33,44,55}
s2 = {22, 66, 77, 88,99, 33}

s1.union(s2)
```

```python
# 'intersection'

s1 = {11,22,33,44,55}
s2 = {22, 66, 77, 88,99, 33}

s1.intersection(s2)
```

```python
# 'difference'

s1 = {11,22,33,44,55}
s2 = {22, 66, 77, 88,99, 33}

s1.difference(s2)    # s1 - s2    # 11, 44, 55
```

```python
s2.difference(s1)            # s2 - s1   #  66,77,88,99
```

```python
# 'symmetric_difference'

s1 = {11,22,33,44,55}
s2 = {22, 66, 77, 88,99, 33}

s1.symmetric_difference(s2)
```

```python
# 'add'

s1 = {11,22,33,44,55}

s1.add(99)
print(s1)
```

```python
# Deletion 
```

```python
# pop()

s1 = {11,22,33,44,55}
s1.pop()

print(s1)
```

```python
# remove()

s1 = {11,22,33,44,55}
s1.remove( 22 )

print(s1)
```

```python
# discard()

s1 = {11,22,33,44,55}
s1.discard( 99 )

print(s1)
```

```python
# update
```

```python
s1 = {11,22,33,44,55}

s1.update({77, 88, 33 })

print(s1)
```

```python
# 'intersection_update' 

s1 = {11 , 22, 33, 44, 55, 66}
s2 = { 33, 44, 99, 88 }

s1.intersection_update(s2)

print(s1)
print(s2)
```

```python
# 'difference_update'

s1 = {11 , 22, 33, 44, 55, 66}
s2 = { 33, 44, 99, 88 }

s1.difference_update(s2)

print(s1)
print(s2)
```

```python
# 'symmetric_difference_update'

s1 = {11 , 22, 33, 44, 55, 66}
s2 = { 33, 44, 99, 88 }

s1.symmetric_difference_update(s2)
print(s1)
```
